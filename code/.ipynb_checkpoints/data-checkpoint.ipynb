```python
import torch.nn as nn
class LSAAttention(nn.Module):
    """
    LSA: self attention within a group
    """
    def __init__(self, dim, num_heads=8, qkv_bias=False, qk_scale=None, attn_drop=0., proj_drop=0., ws=9):
        assert ws != 1
        super(LSAAttention, self).__init__()
        assert dim % num_heads == 0, f"dim {dim} should be divided by num_heads {num_heads}."

        self.dim = dim
        self.num_heads = num_heads
        head_dim = dim // num_heads
        self.scale = qk_scale or head_dim ** -0.5

        self.qkv = nn.Linear(dim, dim * 3, bias=qkv_bias)
        self.attn_drop = nn.Dropout(attn_drop)
        self.proj = nn.Linear(dim, dim)
        self.proj_drop = nn.Dropout(proj_drop)
        self.ws = ws

    def forward(self, x, H, W):
        b, c, h, w, d = x.shape
        x = x.reshape(b, -1, c)
        print("x: ", x.shape)
        
        B, N, C = x.shape
        h_group, w_group = H // self.ws, W // self.ws

        total_groups = h_group * w_group
        print("total_groups: ", total_groups)

        x = x.reshape(B, h_group, self.ws, w_group, self.ws, C).transpose(2, 3)
        print("x: ", x.shape)

        qkv = self.qkv(x).reshape(B, total_groups, -1, 3, self.num_heads, C // self.num_heads).permute(3, 0, 1, 4, 2, 5)
        print("qkv: ", qkv.shape)
        # B, hw, ws*ws, 3, n_head, head_dim -> 3, B, hw, n_head, ws*ws, head_dim
        q, k, v = qkv[0], qkv[1], qkv[2]  # B, hw, n_head, ws*ws, head_dim
        print("q: ", q.shape)
        print("k: ", k.shape)
        print("v: ", v.shape)
        attn = (q @ k.transpose(-2, -1)) * self.scale  # B, hw, n_head, ws*ws, ws*ws
        print("attn: ", attn.shape)
        attn = attn.softmax(dim=-1)
        attn = self.attn_drop(
            attn)  # attn @ v-> B, hw, n_head, ws*ws, head_dim -> (t(2,3)) B, hw, ws*ws, n_head,  head_dim
        attn = (attn @ v).transpose(2, 3).reshape(B, h_group, w_group, self.ws, self.ws, C)
        print("attn: ", attn.shape)
        x = attn.transpose(2, 3).reshape(B, N, C)
        print("x: ", x.shape)
        x = self.proj(x)
        print("x: ", x.shape)
        x = self.proj_drop(x)
        
        x = x.reshape(b, c, h, w, d)
        return x

x = torch.rand(1, 32, 9, 11, 9)
net = LSAAttention(32)
y = net(x, 9*11, 9)
```

```python
import torch.nn as nn
class GSAAttention(nn.Module):
    """
    GSA: using a  key to summarize the information for a group to be efficient.
    """
    def __init__(self, dim, num_heads=8, qkv_bias=False, qk_scale=None, attn_drop=0., proj_drop=0., sr_ratio=1):
        super().__init__()
        assert dim % num_heads == 0, f"dim {dim} should be divided by num_heads {num_heads}."

        self.dim = dim
        self.num_heads = num_heads
        head_dim = dim // num_heads
        self.scale = qk_scale or head_dim ** -0.5

        self.q = nn.Linear(dim, dim, bias=qkv_bias)
        self.kv = nn.Linear(dim, dim * 2, bias=qkv_bias)
        self.attn_drop = nn.Dropout(attn_drop)
        self.proj = nn.Linear(dim, dim)
        self.proj_drop = nn.Dropout(proj_drop)

        self.sr_ratio = sr_ratio
        if sr_ratio > 1:
            self.sr = nn.Conv2d(dim, dim, kernel_size=sr_ratio, stride=sr_ratio)
            self.norm = nn.LayerNorm(dim)

    def forward(self, x, H, W):
        b, c, h, w, d = x.shape
        x = x.reshape(b, -1, c)
        print("x: ", x.shape)
        
        B, N, C = x.shape
        q = self.q(x).reshape(B, N, self.num_heads, C // self.num_heads).permute(0, 2, 1, 3)
        print("q: ", q.shape)
        if self.sr_ratio > 1:
            x_ = x.permute(0, 2, 1).reshape(B, C, H, W)
            x_ = self.sr(x_).reshape(B, C, -1).permute(0, 2, 1)
            x_ = self.norm(x_)
            kv = self.kv(x_).reshape(B, -1, 2, self.num_heads, C // self.num_heads).permute(2, 0, 3, 1, 4)
        else:
            kv = self.kv(x).reshape(B, -1, 2, self.num_heads, C // self.num_heads).permute(2, 0, 3, 1, 4)
        k, v = kv[0], kv[1]
        print("k: ", k.shape)
        print("v: ", v.shape)

        attn = (q @ k.transpose(-2, -1)) * self.scale
        attn = attn.softmax(dim=-1)
        attn = self.attn_drop(attn)
        print("attn: ", attn.shape)

        x = (attn @ v).transpose(1, 2).reshape(B, N, C)
        print("x: ", x.shape)
        x = self.proj(x)
        x = self.proj_drop(x)
        print("x: ", x.shape)

        x = x.reshape(b, c, h, w, d)
        return x

# x = torch.rand(1, 32, 9, 11, 9)
# net = GSAAttention(32)
# y = net(x, 9*11, 9)
```

```python
class FuseBlock(nn.Module):
    def __init__(self, in_c):
        super(FuseBlock, self).__init__()
        self.conv_q = nn.Conv3d(in_c, in_c, kernel_size=3, stride=1, padding=1)
        self.conv_k = nn.Conv3d(in_c, in_c, kernel_size=3, stride=1, padding=1)
        self.conv_v = nn.Conv3d(in_c, in_c, kernel_size=3, stride=1, padding=1)
        self.ln = nn.LayerNorm(normalized_shape=(32, 9, 11, 11))
    
    def forward(self, x):
        q = self.conv_q(x)
        print("q: ", q.shape)
        k = self.conv_k(x)
        print("k: ", k.shape)
        v = self.conv_v(x)
        
        f = q @ k.transpose(-2, -1)
        print("f: ", f.shape)
        f = self.ln(f)

        out = f@v
        
        return out

# x = torch.rand(1, 32, 9, 11, 9)

# net = FuseBlock(32)
# y = net(x)
# y.shape
```

```python
# 计算雅可比矩阵以及统计负数位置数目
import SimpleITK as sitk
import os
import numpy as np
import torch

def check_jacobian_positive(J):
    det_J = np.linalg.det(J)
    return det_J > 0
    
def Get_Ja(displacement):

    D_x = (displacement[:,1:,:-1,:-1,:] - displacement[:,:-1,:-1,:-1,:])

    D_y = (displacement[:,:-1,1:,:-1,:] - displacement[:,:-1,:-1,:-1,:])

    D_z = (displacement[:,:-1,:-1,1:,:] - displacement[:,:-1,:-1,:-1,:])

    

    D1 = (D_x[...,0]+1)*( (D_y[...,1]+1)*(D_z[...,2]+1) - D_z[...,1]*D_y[...,2])

    D2 = (D_x[...,1])*(D_y[...,0]*(D_z[...,2]+1) - D_y[...,2]*D_z[...,0])

    D3 = (D_x[...,2])*(D_y[...,0]*D_z[...,1] - (D_y[...,1]+1)*D_z[...,0])

    #D = np.abs(D1-D2+D3)

    return D1-D2+D3


# path_m = "/home/hzy/projects/MICCAI-24/Results_LPBA40_Test"
# path_m = "/home/hzy/projects/MICCAI-24/Results_MRCT_Test(CT2MR)"
# path_m = '/home/hzy/projects/MICCAI-24/Results_L2R_Test'
path_m = '/home/hzy/projects/MICCAI-24/Test_Result_OASIS'
num = os.listdir(path_m)
print("begin...")
for m in num:
    p = file = os.path.join(path_m, m, "save_image")

    file_list = os.listdir(p)
    y = 0
    x = []
    for f in file_list:
        if f[0] =='O':
            # print(f)
            file = os.path.join(p, f, "flow.nii.gz")
            ref_img = sitk.ReadImage(file)
            img = sitk.GetArrayFromImage(ref_img)[np.newaxis, ...]
            # jacobian_matrix = np.empty((112, 192, 192, 3, 3))  # 创建一个空的雅可比矩阵
            # print(img.shape)
            img = Get_Ja(img)
            # img = check_jacobian_positive(img)
            b,c,h,d = img.shape
      
            # print("img: ", img.shape)
        
            # print(img.shape)
            # print(jacobian_matrix)
            n = (img<0)
            s = n.astype(int)
            # t = s.sum() / (144*176*144*3)
            
            x.append(s.sum() / (b*c*h*d))
  # 
            # print(s.sum())
    print("method: ", m, np.mean(x), np.std(x))
    # print("mean: ", np.mean(x))
print("end !!!")
```

```python
# 形变场可视化
import matplotlib.pyplot as plt
import SimpleITK as sitk
import numpy as np


def grid2contour(grid, title):
    '''
    grid--image_grid used to show deform field
    type: numpy ndarray, shape： (h, w, 2), value range：(-1, 1)
    '''
    assert grid.ndim == 3
    x = np.arange(-1, 1, 2.0 / grid.shape[1])
    y = np.arange(-1, 1, 2.0 / grid.shape[0])
    X, Y = np.meshgrid(x, y)
    Z1 = grid[:, :, 0] + 2  # remove the dashed line
    Z1 = Z1[::-1]  # vertical flip
    Z2 = grid[:, :, 1] + 2

    plt.figure()
    plt.contour(X, Y, Z1, 15, levels=25, colors='k') #改变levels的值，可以改变形变场的密集程度
    plt.contour(X, Y, Z2, 15, levels=25, colors='k')
    plt.xticks(()), plt.yticks(())  # remove x, y ticks
    plt.title(title)
    plt.show()


def show_grid():
    img = sitk.ReadImage("/home/hzy/projects/MICCAI-24/Test_Result_Lbpa/Demo/save_image/S06/144_flow.nii.gz")
    img_arr = sitk.GetArrayFromImage(img)
    img_shape = img_arr.shape

    # 起点、终点、步长（可为小数）
    x = np.arange(-1, 1, 2 / img_shape[1])
    y = np.arange(-1, 1, 2 / img_shape[0])
    X, Y = np.meshgrid(x, y)
    regular_grid = np.stack((X, Y), axis=2)
    grid2contour(regular_grid, "regular_grid")

    rand_field = np.random.rand(*img_shape[:2], 2)  # 参数前加*是以元组形式导入
    rand_field_norm = rand_field.copy()
    rand_field_norm[:, :, 0] = rand_field_norm[:, :, 0] * 2 / img_shape[1]
    rand_field_norm[:, :, 1] = rand_field_norm[:, :, 1] * 2 / img_shape[0]
    sampling_grid = regular_grid + rand_field_norm
    grid2contour(sampling_grid, "sampling_grid")

    img_arr[..., 0] = img_arr[..., 0] * 2 / img_shape[1]
    img_arr[..., 1] = img_arr[..., 1] * 2 / img_shape[0]
    img_grid = regular_grid + img_arr
    grid2contour(img_grid, "img_grid")


if __name__ == "__main__":
    show_grid()
    print("end")
```

```python
# fix和 move 图像的残差
import SimpleITK as sitk
import os
import numpy as np
import torch
import matplotlib.pyplot as plt
import torch.nn.functional as F

class NCCLoss:

    def __init__(self, window=(9, 9, 9)):
        self.win = window
     
    def __call__(self, I, J, weight=False):
        sum_weight = torch.ones((1, 1, self.win[0], self.win[1], self.win[2]))

        I2 = I*I
        J2 = J*J
        IJ = I*J

        # compute local sum
        padding = [ int((k)/2) for k in self.win ]
        I_sum = F.conv3d(I, sum_weight, padding=padding)
        J_sum = F.conv3d(J, sum_weight, padding=padding)
        I2_sum = F.conv3d(I2, sum_weight, padding=padding)
        J2_sum = F.conv3d(J2, sum_weight, padding=padding)
        IJ_sum = F.conv3d(IJ, sum_weight, padding=padding)

        # cross correlation
        win_size = self.win[0] * self.win[1] * self.win[2]
        I_u = I_sum/win_size
        J_u = J_sum/win_size


        cross = IJ_sum - J_u*I_sum - I_u*J_sum + I_u*J_u*win_size
        I_var = I2_sum - 2*I_u*I_sum + I_u*I_u*win_size
        J_var = J2_sum - 2*J_u*J_sum + J_u*J_u*win_size

        cc = cross*cross / (I_var*J_var + 1e-5)

        #weight in different slice
        if weight:
            cc = torch.mean(cc, dim=[0,1,3,4])
            w = torch.zeros(cc.size(0), dtype=cc.dtype, device = cc.device)
            for i in range(0, w.size(0), 4):
                w[i] = 1.0/w.size(0)*4.0
            cc = cc*w

        return cc.mean()

fix = '/home/hzy/projects/MICCAI-24/Results_LPBA40_Test/ProExtension/save_image/fixed.nii.gz'
mov = '/home/hzy/projects/MICCAI-24/Results_LPBA40_Test/ModeT/save_image/S03/warped.nii.gz'
NCC = NCCLoss()


fix_img = sitk.GetArrayFromImage(sitk.ReadImage(fix))
mov_img = sitk.GetArrayFromImage(sitk.ReadImage(mov))
print(fix_img.shape)
# fix_img = sitk.GetArrayFromImage(sitk.ReadImage(fix)).reshape(192, 192, 96)
# mov_img = sitk.GetArrayFromImage(sitk.ReadImage(mov)).reshape(192, 192, 96)
print(fix_img.shape)

mse = np.abs(fix_img[71, :, :]-mov_img[71, :, :]).mean()
ncc = NCC(torch.from_numpy(fix_img[71, :, :]).unsqueeze(0).unsqueeze(0), torch.from_numpy(mov_img[71, :, :]).unsqueeze(0).unsqueeze(0))
print("mse: ", mse)
print("ncc: ", ncc)
res_img = fix_img[71, :, :]-mov_img[71, :, :]
res_img = np.abs(res_img)

gamma = 0.5
# res_img = np.power(np.abs(res_img), gamma)
# res_img = (gamma_corrected_residual - np.min(gamma_corrected_residual)) / (np.max(gamma_corrected_residual) - np.min(gamma_corrected_residual)) * 255


plt.figure(figsize=(20,12))
plt.imshow(res_img, cmap='jet')
plt.colorbar()
plt.axis('off')
plt.show()
```

```python
import itk
import SimpleITK as sitk
import os
import numpy as np
import torch
import matplotlib.pyplot as plt
import torch.nn.functional as F

# 读取图像
fix = '/home/hzy/projects/MICCAI-24/Results_LPBA40_Test/ProExtension/save_image/fixed.nii.gz'
mov = '/home/hzy/projects/MICCAI-24/Results_LPBA40_Test/ProExtension/save_image/S03/warped.nii.gz'
fixed_image = itk.imread(fix)
moving_image = itk.imread(mov)

# 创建用于重采样的变换，这里使用 IdentityTransform 作为示例
transform = itk.IdentityTransform[itk.D, fixed_image.GetImageDimension()].New()

# 创建重采样过滤器
resample_filter = itk.ResampleImageFilter.New(Input=moving_image, Transform=transform, UseReferenceImage=True, ReferenceImage=fixed_image)
resample_filter.SetDefaultPixelValue(2)
resample_filter.Update()

# 获取重采样后的图像
resampled_image = resample_filter.GetOutput()

# 创建 CheckerBoardImageFilter
checker_board_filter = itk.CheckerBoardImageFilter.New(Input1=fixed_image, Input2=resampled_image)

# 设置 CheckerBoard 的参数，例如棋盘块的数量
checker_pattern = itk.FixedArray   # 3D 图像，设置每个方向上的格子数目
checker_board_filter.SetCheckerPattern(6)  # 4x4 的棋盘块

# 生成棋盘图像
checker_board_filter.Update()

# 获取输出图像
checkerboard_image = checker_board_filter.GetOutput()

# 保存或显示图像
# itk.imwrite(checkerboard_image, 'checkerboardImage.nii')

# 如果你想显示图像，可以使用 itk.show
# itk.show(checkerboard_image, title='Checkerboard Image')

print(checkerboard_image.shape)
# 绘制棋盘格图像，使用不同的颜色区分两幅图像
plt.figure(figsize=(20, 12))
plt.imshow(checkerboard_image[71, :, :], cmap='gray', interpolation='nearest', alpha=1.0)  # 选择适当的切片进行显示
# plt.imshow(checkerboard_image[71, :, :], cmap='jet', interpolation='nearest', alpha=0.5)  # 使用 inferno 颜色图来显示第二幅图像
plt.axis('off')  # 关闭坐标轴
plt.title('Checkerboard Image')
plt.colorbar()
plt.show()
```
